# GHO OData API 
The GHO portal provides a simple query interface to the World Health Organization's data and statistics content, using OData (Open Data Protocol).

You can find below a set with the most comprehensive examples. 

*More information about the Odata protocol can be found [here](https://www.odata.org/documentation/odata-version-2-0/uri-conventions/).

## References
* https://www.who.int/data/gho/info/gho-odata-api

```python
# Import Required Libraries
import pandas as pd
import requests
import plotly.express as px
import plotly.graph_objects as go

# WHO GHO API Base URL
API_BASE_URL = "https://ghoapi.azureedge.net/api"
```

# Available Dimensions

```python
# Fetch available dimensions
response = requests.get(f"{API_BASE_URL}/Dimension")
dimensions = pd.DataFrame(response.json()["value"])
print(f"Available Dimensions: {len(dimensions)}")
```

```python
dimensions.head(50)
```

```python
dimensions.to_csv("dimensions.csv", index=False)
```

## Countries

```python
# Fetch countries
response = requests.get(f"{API_BASE_URL}/DIMENSION/COUNTRY/DimensionValues")
countries = pd.DataFrame(response.json()["value"])
print(f"Total Countries: {len(countries)}")
```

```python
countries
```

```python
countries[["Code", "Title"]].head(20)
```

## WHO Regions

```python
# Fetch WHO regions
response = requests.get(f"{API_BASE_URL}/DIMENSION/REGION/DimensionValues")
regions = pd.DataFrame(response.json()["value"])
print(f"WHO Regions: {len(regions)}")
```

```python
regions
```

```python
regions[["Code", "Title"]]
["AFR", "AMR", "EMR","EUR", "NC", "GLOBAL", "SEAR", "WPR"]
```

## SDG Health Indicators

```python
response = requests.get(f"{API_BASE_URL}/Indicator/")
indicators = pd.DataFrame(response.json()["value"])
print(f"Total Indicators: {len(indicators)}")
```
